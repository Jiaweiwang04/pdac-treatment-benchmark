```python
import csv
from pathlib import Path

repo = Path.cwd().resolve()
if repo.name == 'notebooks':
    repo = repo.parent
for name in ['file_inventory.csv', 'field_inventory.csv', 'table_relationships.csv', 'feasibility_summary.csv']:
    path = repo / 'reports' / 'tables' / name
    with path.open(encoding='utf-8', newline='') as f:
        rows = list(csv.DictReader(f))
    print(name, 'rows=', len(rows))
    for row in rows[:5]:
        public = {k: row[k] for k in list(row)[:6]}
        print(public)
    print()
```
