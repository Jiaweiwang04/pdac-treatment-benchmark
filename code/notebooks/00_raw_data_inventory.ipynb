# BPC PANC Raw Data Read-Only Audit

This notebook runs `code/scripts/audit_raw_data.py` and inspects aggregate outputs only. Do not display patient-level records here.


```python
from pathlib import Path
import subprocess, sys

def find_repo_root(start):
    for p in [start, *start.parents]:
        if (p / 'data').exists() and (p / 'code' / 'scripts' / 'audit_raw_data.py').exists():
            return p
    raise RuntimeError('repo root not found')
repo = find_repo_root(Path.cwd().resolve())
script = repo / 'code' / 'scripts' / 'audit_raw_data.py'
subprocess.run([sys.executable, str(script), '--repo-root', str(repo)], check=True)
```

```python
import csv
from pathlib import Path

def find_repo_root(start):
    for p in [start, *start.parents]:
        if (p / 'data').exists() and (p / 'code' / 'scripts' / 'audit_raw_data.py').exists():
            return p
    raise RuntimeError('repo root not found')
repo = find_repo_root(Path.cwd().resolve())
for name in ['file_inventory.csv', 'field_inventory.csv', 'table_relationships.csv', 'feasibility_summary.csv']:
    path = repo / 'code' / 'results' / 'data_audit' / 'tables' / name
    with path.open(encoding='utf-8', newline='') as f:
        rows = list(csv.DictReader(f))
    print(name, 'rows=', len(rows))
    for row in rows[:5]:
        public = {k: row[k] for k in list(row)[:6]}
        print(public)
    print()
```
